Question 3: The ColumnTransformer lets you apply different preproccesing steps to different types of columns at the same time. Wrapping each of those in a pipeline keeps the steps bundled together instead of being done separately by hand. This helps prevent data leakage because when you fit the pipeline only on the training data, it learns things like the mean and standard deviation for scaling or the categories for encoding only from that training data, and then applies those same learned values to the test data instead of recalculating anything from it.
