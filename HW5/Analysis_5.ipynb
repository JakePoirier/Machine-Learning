Implicit missing values are a problem because the data looks complete when it isn’t. info() showed 768 values in every column, but describe() showed a minimum of 0 for glucose, blood pressure, skin thickness, insulin, and BMI, which isn’t possible for a living patient. When I swapped those zeros for blanks, I found 5 in glucose, 35 in blood pressure, 227 in skin thickness, 374 in insulin, and 11 in BMI. If I had left them, the models would have treated them as real measurements. I filled them with the median instead of deleting rows because insulin is missing for 374 of 768 patients, so deleting would throw away almost half the data. I used the median over the average because insulin goes up to 846, so a few extreme patients pull the average up.

2. In hard voting, each model gives a yes or no and the answer with the most votes wins. In soft voting, each model gives a chance of diabetes and the team averages them, so a model that is very sure counts for more. Hard voting got 74.0% and soft voting got 72.7%, a difference of only 2 patients out of 154. Well-calibrated probabilities matter because the average only works if the numbers mean the same thing across models. If one model says 90% when it is only right about 70% of the time, it outweighs the others without earning it.

3. Each tree in Bagging is trained on a random sample picked with replacement, so it leaves out about a third of the training patients. Those are its out-of-bag patients. To get the OOB score, each patient is graded using only the trees that never saw them. It replaces cross-validation because it comes from a single training run, while 5-fold cross-validation trains the model 5 times. My OOB score was 0.7557 and the test accuracy was 0.7208.

4. Standard Bagging lets every tree look at all 8 features at every split, so a strong feature like glucose ends up at the top of almost every tree and the trees make similar mistakes. A Random Forest only lets each split look at a small random group of features, so the trees learn different patterns and their mistakes cancel out more when they vote. That lowers the variance and helps on new patients. Bagging got 72.1% on the test set and the Random Forest got 77.3%, with diabetic recall going from 0.52 to 0.61.

5. AdaBoost trains its stumps one after another. After each stump, it raises the weight of the patients that stump got wrong, so the next stump focuses on the hard cases, and the more accurate stumps get more say in the final vote. My final train accuracy was 81.4% and test accuracy was 76.0%, a gap of about 5 points, so there is a sign of overfitting. The best test accuracy was 76.6% at 21 stumps, so the extra stumps didn’t help. To reduce it, I could lower the learning rate, use fewer stumps.
