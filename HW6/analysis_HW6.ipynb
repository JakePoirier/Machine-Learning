1. With the full dataset, 123 principal components kept 95% of the information, out of the original 4,096, which is about 3% of the size. When I ran it on the training faces only in Task 5, it needed 111. At k = 10 the faces are very blurry. You can tell it’s a face, and the eyes and eyebrows are in roughly the right place, but the nose, mouth, and skin texture are smeared together, and the glasses are gone completely. It looks like an average face instead of the actual person. At k = 50 the person is recognizable but still soft, and at k = 100 the glasses and facial hair come back. At k = 200 it first looks the same as the original, and at k = 300 I can’t tell the difference. The error numbers match this. The mean squared error went from 0.00577 at k = 10 to 0.00214 at k = 50, 0.00109 at k = 100, 0.00040 at k = 200, and 0.00013 at k = 300, so most of the improvement happens early and the later components add very little.

2. The first eigenface is a bright, blurry whole face, so it captures the overall face shape and brightness that most of the faces share. PC 2 and PC 3 are also whole face patterns, but they look like lighting, with one side of the face brighter or darker. Further down, the eigenfaces pick out specific things like glasses frames, mouth and smile shapes, the nose, and beard areas. The later components look more like fine grained noise because the big patterns that every face shares are already taken by the early ones. What’s left over is small differences that only show up in a few pictures, like tiny shadows and individual pixel changes. By PC 15 each one holds only about 1% of the information, and the patterns were already finer and more scattered than the smooth first few.

3. PCA has to be fit on the training split only, because the test faces are supposed to act like new faces the model has never seen. If I fit PCA on the whole dataset first, the test faces help decide what the components look like, so information from the test set leaks into the training. The model is then graded on pictures that already shaped its input, so the test accuracy would look better than it really is, and I wouldn’t know how it would do on a new face. PCA on all 400 faces needed 123 components for 95%, but on the training faces only it needed 111, so the two versions don’t build exactly the same components.

4. Logistic Regression got 96.25% on PCA features, the RBF SVC got 93.75%, and the soft voting ensemble got 95.0%. The curved boundaries and the team of models didn’t give a real gain. The differences are 1 to 3 faces out of 80, so they’re small enough to be luck. Logistic Regression was actually the best, and it tied with the raw pixel version at 96.25%. Classical models hit a ceiling because they treat the picture as one long list of 4,096 numbers, so they don’t know which pixels are next to each other. A face that is shifted a little or lit differently changes a lot of the pixel values even though it’s the same person, and PCA looks at the whole picture at once, so it can’t pick up on local shapes like an eye or an edge. There’s also very little data to learn from, only 8 training pictures per person.

5. In Task 8 the accuracy was almost the same for all four models, between 93.75% and 96.25%, but the training time was not. Logistic Regression on the raw pixels took 29.73 seconds, while the PCA models took 0.30 to 2.00 seconds, so PCA was 15 to 100 times faster with no loss in accuracy. That shows PCA is good at shrinking the data, but it isn’t enough for hard computer vision tasks. It is a global, straight line method, so it still treats the picture as a list of numbers and can’t learn local features like an eye or an edge wherever they show up. It worked because the faces are all the same size, centered, and from a small dataset of 400 pictures. With larger and messier pictures these models would fall apart. Those tasks need methods that look at small patches of the picture and learn the features themselves, like neural networks built for images.
